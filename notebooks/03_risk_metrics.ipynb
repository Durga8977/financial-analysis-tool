# Risk Metrics Notebook

Calculate and analyze key risk metrics:
- Volatility
- Sharpe Ratio
- Sortino Ratio
- Maximum Drawdown
- Value at Risk
- Correlation Analysis

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import requests

API_BASE_URL = 'http://localhost:5000/api'
PORTFOLIO_ID = 1
```

## 1. Portfolio Risk Metrics

```python
# Get portfolio risk metrics
response = requests.get(f'{API_BASE_URL}/risk/portfolio-metrics/{PORTFOLIO_ID}?days=90')
risk_data = response.json()

print("Portfolio Risk Analysis (90-day):")
print(f"Volatility: {risk_data.get('volatility', 0):.4f}")
print(f"Sharpe Ratio: {risk_data.get('sharpe_ratio', 0):.4f}")
print(f"Sortino Ratio: {risk_data.get('sortino_ratio', 0):.4f}")
print(f"Max Drawdown: {risk_data.get('max_drawdown', 0):.4f}")
print(f"Value at Risk (95%): {risk_data.get('value_at_risk_95', 0):.4f}")
print(f"Value at Risk (99%): {risk_data.get('value_at_risk_99', 0):.4f}")
```

## 2. Individual Holding Risk Metrics

```python
# Get risk metrics for a holding
holding_id = 1  # Change to your holding ID

response = requests.get(f'{API_BASE_URL}/risk/holding-metrics/{holding_id}?days=90')
holding_risk = response.json()

print(f"Risk Metrics for {holding_risk.get('symbol')}:")
print(f"Volatility: {holding_risk.get('volatility', 0):.4f}")
print(f"Sharpe Ratio: {holding_risk.get('sharpe_ratio', 0):.4f}")
print(f"Max Drawdown: {holding_risk.get('max_drawdown', 0):.4f}")
print(f"Value at Risk (95%): {holding_risk.get('value_at_risk_95', 0):.4f}")
```

## 3. Correlation Analysis

```python
# Get correlation matrix
symbols = ['AAPL', 'MSFT', 'GOOGL', 'TSLA']
asset_types = {s: 'stock' for s in symbols}

response = requests.post(f'{API_BASE_URL}/risk/correlation?days=90',
                         json={'symbols': symbols, 'asset_types': asset_types})
correlation_data = response.json()['correlation_matrix']

df_corr = pd.DataFrame(correlation_data).T
print("Correlation Matrix:")
print(df_corr)
```

```python
# Visualize correlation heatmap
fig = go.Figure(data=go.Heatmap(z=df_corr.values, x=df_corr.columns, y=df_corr.index,
                                  colorscale='RdBu', zmid=0))
fig.update_layout(title='Asset Correlation Heatmap')
fig.show()
```

## 4. Risk Interpretation Guide

### Sharpe Ratio
- < 1: Sub-optimal
- 1-2: Good
- 2-3: Very Good
- > 3: Excellent

### Sortino Ratio
- Similar to Sharpe but focuses on downside risk
- Higher is better

### Maximum Drawdown
- Largest peak-to-trough decline
- Lower (closer to 0) is better

### Volatility
- Standard deviation of returns
- Higher volatility = higher risk

### Value at Risk (VaR)
- 95% VaR: 95% chance of not losing more than this
- 99% VaR: 99% chance of not losing more than this
